```python
import pandas as pd

df = pd.read_csv("D:/IIITD/# SEM/IIA/house-recommender/datasources/datasource1_flats/kolkata.csv")
import ast

# Convert the column from string to dictionary
df["location"] = df["location"].apply(lambda x: ast.literal_eval(x) if pd.notna(x) else {})

# Convert dict to multiple columns
location_df = pd.json_normalize(df["location"])

# Combine with original dataframe (drop the old column if you want)
df = pd.concat([df.drop(columns=["location"]), location_df], axis=1)



```

```python
# Remove all non-numeric characters and convert to int
df["AREA"] = df["AREA"].astype(str).str.extract(r"(\d+)").astype(float).astype("Int64")
df["AREA"][0:5]
```
