```python
# Load model directly
from transformers import AutoTokenizer, AutoModelForCausalLM

model_name = "openai-community/gpt2"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(model_name)
```

```python
def describe_property(data_dict):
    prompt = f"You are Real Estate expert. Your role is to give all of the options present for a particular user. You are given the complete data about that property. Your task is to convert this property info into a natural summary which includes all the data present in this dictionary: {data_dict}"
    inputs = tokenizer(prompt, return_tensors="pt")
    outputs = model.generate(**inputs)
    return tokenizer.decode(outputs[0], skip_special_tokens=True)

data = {
    'preferred_location': 'Navi Mumbai',
    'preferred_location2': 'Sector 16A Nerul',
    'budget': '1.5 L',
    'bedrooms': 4.0,
    'coordinate': "{'LATITUDE': '19.035977', 'LONGITUDE': '73.008537'}",
    'furnish_type': 1,
    'area': 3000,
    'bldg_name': 'Sea Breeze Tower',
    'price': '1.5 L',
    'hospital': 7,
    'Gym': 20,
    'Swimming Pool': 5,
    'Club House': 2,
    "Children's Play Area": 5,
    'School Nearby': 16,
    'nursery': 13,
    'Park Nearby': 2,
    'review_summary': "This residential society in Nerul, Navi Mumbai, located along Palm Beach Road facing the creek, is highly acclaimed for its premium infrastructure, exceptional management, and resident-centric amenities. It features spacious gardens, excellent security, well-maintained buildings, and a clean, pollution-free ambiance. Key highlights include a large children’s park, a Ganesha temple, regular cultural events, a pooja hall, and convenient access to shopping, fresh produce, and restaurants. The society has been recognized with the award 'QUEEN OF TOWERS' and is regarded as the best in Navi Mumbai and the top society on Palm Beach Road. Residents enjoy stunning sunset views and a vibrant, well-run community life.",
    'sentiment': 1
 }
result = describe_property(data)
```
